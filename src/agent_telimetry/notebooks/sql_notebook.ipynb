```python
import psycopg
from dotenv import load_dotenv
import os
import pandas as pd

load_dotenv()
connection_string = os.getenv("DATABASE_STRING")

conn = psycopg.connect(
    connection_string if connection_string else ""
)
```

```python
def run(query):
    
    results = pd.read_sql(query, conn)
    conn.commit()
    return results
```

```python
query = "Select * from llm_calls where status = 'success'"
run(query)
```
