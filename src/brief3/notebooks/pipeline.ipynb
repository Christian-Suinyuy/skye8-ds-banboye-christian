```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../data/raw/support_messages.csv")
df.head()
```

```python
x = df.drop("intent", axis=1)
y = df["intent"]
```

```python
x_train, x_temp, y_train, y_temp = train_test_split(x, y, test_size=0.3, random_state=54, stratify=y)
```

```python
from sklearn.dummy import DummyClassifier
model = DummyClassifier(strategy="most_frequent")
```

```python
vectorizer = TfidfVectorizer(analyzer='char', ngram_range=(1,3))
encoder = OneHotEncoder(handle_unknown="ignore")

encode_columns = ["channel","language_hint","customer_tier"]

column_transform = ColumnTransformer([
    ("vectorizer", vectorizer,"message_text"),
    ("encode", encoder, encode_columns )
], remainder="drop")

pipeline = Pipeline(steps=[
    ("column_transformer", column_transform),
    ("model", model)
])
```
